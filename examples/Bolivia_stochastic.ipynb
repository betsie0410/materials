```python
import mpisppy.utils.sputils as sputils

def scenario_creator(scenario_name):
    sdata = DEMAND_SCENARIOS[scenario_name]

    # ---- scale demand ----
    demand_s = df_base.copy()
    demand_s[0] = sdata["multiplier"] * demand_s[0]

    # ---- Declare Location ----
    process_set = {NG_prod, Oil_prod, PV, WF, Hydro_prod, Biomass_prod}
    bolivia = Location(
        name='BO',
        processes=process_set,
        capacity_factor={PV: solar_bolivia, WF: wind_bolivia[:8760*horizon]},
        demand_factor={Power: demand_s},
        price_factor={NG: ng_price},
        scales=scales,
        label='Bolivia',
        demand_scale_level=2,
        capacity_scale_level=2,
        price_scale_level=2
    )

    # ---- Declare Scenario ----
    first_stage_processes = {PV, WF, Hydro_prod}

    scenario = Scenario(
        name='scenario_full',
        network=bolivia,
        scales=scales,
        expenditure_scale_level=0,
        scheduling_scale_level=2,
        network_scale_level=0,
        demand_scale_level=2,
        capacity_scale_level=2,
        purchase_scale_level=2,
        label='full_case',
        demand={bolivia: {Power: demand_s.max()[0]}},
           # <--- here!
    )

    # ---- Formulate MILP ----
    milp_cost = formulate(
        scenario=scenario,
        constraints={
            Constraints.COST,
            Constraints.INVENTORY,
            Constraints.PRODUCTION,
            Constraints.RESOURCE_BALANCE,
            Constraints.DEMAND,
           Constraints.NETWORK,
           Constraints.EMISSION
       },
       objective=Objective.COST,  # <-- use string instead of Objective.COST
    )

    # ---- Attach scenario probability ----
    #milp_cost = build_model(demand_s)
    sputils.attach_root_node(milp_cost, milp_cost.Capex_network, [milp_cost.Cap_P, milp_cost.X_P])
    milp_cost._mpisppy_probability = sdata["prob"]

    return milp_cost
```

```python
options = {"solver": "gurobi"}
solver_options = {'MIPGap': 0.0005,
                  'Heuristics': 0.2}
scenario_names = list(DEMAND_SCENARIOS.keys())
ef_UI = ExtensiveForm(options, scenario_names, scenario_creator)
```
