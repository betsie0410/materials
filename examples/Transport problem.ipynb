```python
from pyomo.environ import *

m = ConcreteModel()
m.I = Set(initialize=['1', '2'])
m.J = Set(initialize=['1', '2', '3'])
m.T = Set(initialize=['1', '2', '3'])

# Define parameters for cost, supply, and demand
distance = {('1','1'): 2.5,
            ('1','2'): 1.7,
            ('1','3'): 1.8,
            ('2','1'): 2.5,
            ('2','2'): 1.8,
            ('2','3'): 1.4,
            }
supply = {('1','1'): 350,
            ('1','2'): 355,
            ('1','3'): 345,
            ('2','1'): 605,
            ('2','2'): 600,
            ('2','3'): 610,
            }
demand = {('1','1'): 325,
            ('1','2'): 330,
            ('1','3'): 320,
            ('2','1'): 300,
            ('2','2'): 305,
            ('2','3'): 310,
            ('3','1'): 270,
            ('3','2'): 275,
            ('3','3'): 280,
            }

m.x = Var(m.I, m.J, m.T, domain = NonNegativeReals)

def objective_rule(m):
    return sum((90*distance[i,j]/1000)* m.x[i, j, t] for i in m.I for j in m.J for t in m.T)
m.total_cost = Objective(rule = objective_rule, sense=minimize)

def supply_const(m,i,t):
    return sum(m.x[i, j, t] for j in m.J) <= supply[i,t]
m.con1 = Constraint(m.I, m.T, rule = supply_const)

def demand_const(m,j,t):
    return sum(m.x[i, j, t] for i in m.I) >= demand[j,t]
m.con2 = Constraint(m.J, m.T, rule = demand_const)


opt = SolverFactory('gurobi')
results = opt.solve(m, tee = True)

print(results)

m.x.display()
!cat results.yml
```
